# Zomato Dataset Processing using DatasetProcessor

This notebook demonstrates how to import and use the `DatasetProcessor` class from `dataset_processor.py` to **load**, **validate**, **clean**, and **transform** the Zomato restaurant dataset.

## 1. Loading the Dataset

```python
# Import the DatasetProcessor class from our module
from dataset_processor import DatasetProcessor
```

```python
# Load the Zomato dataset using the from_csv classmethod
dp = DatasetProcessor.from_csv('zomato.csv')
```

```python
# Quick peek at the loaded data
dp.df.head()
```

```python
# Check the shape of the dataset
print(f"Dataset shape: {dp.df.shape}")
```

## 2. Dataset Summary

```python
# Get a comprehensive summary of the dataset
dp.summary()
```

## 3. Validation

### 3a. Schema Validation
Check if the dataset contains all the columns we expect.

```python
# Define the columns we expect in the Zomato dataset
expected_columns = [
    'url', 'address', 'name', 'online_order', 'book_table',
    'rate', 'votes', 'phone', 'location', 'rest_type',
    'dish_liked', 'cuisines', 'approx_cost(for two people)',
    'reviews_list', 'menu_item', 'listed_in(type)', 'listed_in(city)'
]

schema_result = dp.validate_schema(expected_columns)
schema_result
```

```python
# Test with a missing column to see validation failure
schema_result_fail = dp.validate_schema(['name', 'rate', 'votes', 'non_existent_column'])
schema_result_fail
```

### 3b. Data Type Validation
Verify that columns have the expected data types.

```python
# Validate data types for key columns
expected_dtypes = {
    'votes': 'int64',
    'name': 'object',
    'online_order': 'object',
    'rate': 'object'  # rate is stored as string like '4.1/5'
}

dtype_result = dp.validate_dtypes(expected_dtypes)
dtype_result
```

## 4. Cleaning

### 4a. Check Missing Values

```python
# Check missing values before cleaning
print("Missing values per column:")
print(dp.df.isnull().sum())
print(f"\nTotal missing values: {dp.df.isnull().sum().sum()}")
```

### 4b. Drop Missing Values
Drop rows where critical columns (`rate`, `cuisines`) have missing data.

```python
# Drop rows where 'rate' or 'cuisines' are missing
dp_cleaned = dp.drop_missing(subset=['rate', 'cuisines'])
print(f"Before: {dp.df.shape[0]} rows")
print(f"After:  {dp_cleaned.df.shape[0]} rows")
```

### 4c. Fill Missing Values
Fill missing values in `dish_liked` and `rest_type` using mode (most frequent value).

```python
# Fill missing 'dish_liked' and 'rest_type' with mode
dp_filled = dp_cleaned.fill_missing(strategy='mode', columns=['dish_liked', 'rest_type'])

# Verify no more nulls in those columns
print(f"dish_liked nulls: {dp_filled.df['dish_liked'].isnull().sum()}")
print(f"rest_type nulls:  {dp_filled.df['rest_type'].isnull().sum()}")
```

```python
# Fill missing numeric column 'votes' with mean
dp_filled = dp_filled.fill_missing(strategy='mean', columns=['votes'])
print(f"votes nulls: {dp_filled.df['votes'].isnull().sum()}")
```

### 4d. Drop Duplicates
Remove duplicate restaurant entries based on `name` and `address`.

```python
# Drop duplicate rows based on name and address
dp_deduped = dp_filled.drop_duplicates(subset=['name', 'address'])
print(f"Before: {dp_filled.df.shape[0]} rows")
print(f"After:  {dp_deduped.df.shape[0]} rows")
```

## 5. Transformation

### 5a. Add New Columns
Extract the numeric rating from the `rate` column (e.g., `'4.1/5'` → `4.1`).

```python
# Add a numeric 'rating' column extracted from 'rate' (e.g., '4.1/5' -> 4.1)
dp_transformed = dp_deduped.add_column(
    'rating',
    lambda df: df['rate'].astype(str).str.replace(r'\s*/\s*5', '', regex=True).apply(
        lambda x: float(x) if x.replace('.', '', 1).isdigit() else None
    )
)

# Verify the new column
dp_transformed.df[['name', 'rate', 'rating']].head(10)
```

```python
# Add a numeric 'cost' column by removing commas from 'approx_cost(for two people)'
dp_transformed = dp_transformed.add_column(
    'cost',
    lambda df: pd.to_numeric(
        df['approx_cost(for two people)'].astype(str).str.replace(',', '', regex=False),
        errors='coerce'
    )
)

# Verify the new column
dp_transformed.df[['name', 'approx_cost(for two people)', 'cost']].head(10)
```

### 5b. Normalize Numeric Columns
Scale `votes`, `rating`, and `cost` to the [0, 1] range using Min-Max scaling.

```python
# Normalize the numeric columns
dp_normalized = dp_transformed.normalize(columns=['votes', 'rating', 'cost'])

# Check the normalized values
dp_normalized.df[['name', 'votes', 'rating', 'cost']].head(10)
```

### 5c. Encode Categorical Columns
Label-encode `online_order`, `book_table`, and `listed_in(type)` for ML-readiness.

```python
# Encode categorical columns
dp_encoded = dp_transformed.encode_categorical(
    columns=['online_order', 'book_table', 'listed_in(type)']
)

# Check encoded values
dp_encoded.df[['name', 'online_order', 'book_table', 'listed_in(type)']].head(10)
```

## 6. Final Summary & Export

```python
# Summary of the transformed dataset
dp_transformed.summary()
```

```python
# Export the cleaned and transformed dataset to a new CSV
dp_transformed.to_csv('zomato_cleaned.csv')
```
